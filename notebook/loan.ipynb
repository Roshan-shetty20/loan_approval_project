```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import LogisticRegression
```

```python
from sklearn.preprocessing import LabelEncoder

loan = df.copy()
categorical_cols = ['education', 'self_employed']

le = LabelEncoder()
for i in categorical_cols:
    loan[i] = le.fit_transform(loan[i])

# Drop unnecessary columns (e.g., ID)
loan = loan.drop('loan_id', axis=1)

# Define Features (F) and Target (T)
F = loan.drop('loan_status', axis=1)
T = loan['loan_status']

# Split the data into Training and Testing sets
Ftrain, Ftest, Ttrain, Ttest = train_test_split(
    F, T,
    test_size=0.2,
    random_state=42
)
```

```python
pipeline = Pipeline([
    ('scaler', StandardScaler()),
    ('model', RandomForestClassifier(
        n_estimators=300,
        random_state=42
    ))
])

pipeline.fit(Ftrain, Ttrain)

Tpred = pipeline.predict(Ftest)
```

```python
AC = accuracy_score(Ttest, Tpred)
AC
```

```python
CM = confusion_matrix(Ttest, Tpred)
CM
```

```python
plt.figure(figsize=(3,4))
sns.heatmap(CM, annot=True, fmt='d', cmap=plt.cm.cool)
plt.title('Confusion Matrix')
plt.show()
```

```python
import joblib

joblib.dump(pipeline, 'loan_approval_pipeline.pkl')

print("Pipeline Saved Successfully!")
```
